```python
from sympy import Function, dsolve, Eq, pi, symbols, sin
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np

tensaoinicial = Function("Vin")
tensaocapacitor = Function("Vc")
tempo = symbols("t")
resistencia = symbols("R")
capacitancia = symbols("C")
edo = Eq(tensaocapacitor(tempo).diff(tempo) + (1/(resistencia*capacitancia))*tensaocapacitor(tempo), (1/(resistencia*capacitancia))*tensaoinicial(tempo))
solucao = dsolve(edo)
print("Equação diferencial ordinária (EDO):")
display(edo)
print("Solução da EDO:")
display(solucao)


tensaoinicial = 2*sin(2*pi*0.2*tempo) + 0.8*sin(2*pi*20*tempo)
print("Tensão inicial:")
display(tensaoinicial)
print("Equação diferencial ordinária (EDO) com tensão inicial:")
edo = Eq(tensaocapacitor(tempo).diff(tempo) + (1/(resistencia*capacitancia))*tensaocapacitor(tempo), (1/(resistencia*capacitancia))*tensaoinicial)
display(edo)
solucao = dsolve(edo)
print("Solução da EDO com tensão inicial:")
display(solucao)
resistencia = int(input("Digite o valor da resistência (R) em ohms: "))
capacitancia = float(input("Digite o valor da capacitância (C) em farads: "))
freq_corte = 1/(2*pi*resistencia*capacitancia)
print("Frequência de corte (Hz):")
display(freq_corte)
tempo = symbols("t")
edo = Eq(tensaocapacitor(tempo).diff(tempo) + (1/(resistencia*capacitancia))*tensaocapacitor(tempo), (1/(resistencia*capacitancia))*tensaoinicial)
solucao = dsolve(edo)
print("Equação diferencial ordinária (EDO) com valores numéricos:")
display(edo)
print("Solução da EDO com valores numéricos:")
display(solucao)

C1 = sp.Symbol('C1')
C1_val = sp.solve(solucao.rhs.subs(tempo, 0), C1)[0]
vc_particular = solucao.rhs.subs(C1, C1_val)
tensaocapacitor_numpy = sp.lambdify(tempo, vc_particular, 'numpy')
tensaoinicial_numpy = sp.lambdify(tempo, tensaoinicial, 'numpy')

tempo_valores = np.linspace(0, 10, 1000)
tensaocapacitor_valores = tensaocapacitor_numpy(tempo_valores)
tensaoinicial_valores = tensaoinicial_numpy(tempo_valores)

plt.figure(figsize=(10, 5))
plt.plot(tempo_valores, tensaoinicial_valores,label="Vin (Entrada)", alpha=0.5)
plt.plot(tempo_valores, tensaocapacitor_valores, label="Vc (Saída)", linewidth=2)
plt.xlabel("Tempo (s)")
plt.ylabel("Tensão (V)")
plt.legend()
plt.grid(True)
plt.show()
```
